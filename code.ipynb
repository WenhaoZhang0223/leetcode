```python
# 滑动窗口    计算过程与两端点的 区间区间区间  之间有联系（连续子数组， 连续子串）
# 双指针      计算过程仅仅与两端点的 位置  之间有联系（如两数之和，三数之和）
# the minimum size subarray sum 
class Solution:
    def minSubArrayLen(self, s: int, nums: list[int]) -> int:
        l = len(nums)
        left = 0
        right = 0
        min_len = float('inf')
        cur_sum = 0 # current sum of the window
        
        while right < l:
            cur_sum += nums[right]
            
            while cur_sum >= s: # if the current sum is bigger than the target
                min_len = min(min_len, right - left + 1)
                cur_sum -= nums[left]
                left += 1
            
            right += 1
        
        return min_len if min_len != float('inf') else 0
```

```python
# 螺旋矩阵
class Solution:
    def generateMatrix(self, n: int) -> list[list[int]]:
        if n < 0:
            return []
        
        # 初始化 N x N 的矩阵
        matrix = [[0] * n for _ in range(n)]
        
        # 初始化边界
        left, right, bottom, top = 0, n - 1, n - 1, 0
        # global 填充的数字
        num = 1 
        
        # 填数字循环
        while top <= bottom and left <= right:
        
            # 从左到右
            for i in range(left, right + 1):
                matrix[top][i] = num
                num += 1
            top += 1
            
            # 从上到下
            for i in range(top, bottom + 1):
                matrix[i][right] = num
                num += 1
            right -= 1
            
            # 从右到左
            for i in range(right, left - 1, -1):
                matrix[bottom][i] = num
                num += 1
            bottom -= 1
            # 从下到上
            for i in range(bottom, top - 1, -1):
                matrix[i][left] = num
                num += 1
            left += 1
        # 返回matrix
        return matrix
```

```python
# 二分查找， 左闭右开
class Solution:
    def search(self, nums:list[int], target:int) -> int:
        left = 0
        right = len(nums)
        
        while left < right:
            middle = left + (right - left) // 2
            
            if nums[middle] > target:
                right = middle
            elif nums[middle] < target:
                left = middle + 1 # +1 是因为 middle 已经被检查过了
            else: # nums[middle] == target
                return middle
        return -1
                
```

```python
# 移除元素   双指针

class Solution:
    def removeElement(self, nums: list[int], val: int) -> int:
        slow = 0
        fast = 0
        size = len(nums)
        while fast < size:
            if nums[fast] != val:
                nums[slow] = nums[fast]
                slow += 1
            fast += 1
        return slow
```

```python
# 两个数组的交集

#  字典和集合
class Solution:
    def intersection(self, nums1: list[int], nums2: list[int]) -> list[int]:
        # Use a dictionary to store the elements of the first array
        table = {}
        for num in nums1:
            if num in table:
                table[num] += 1
            else:
                table[num] = 1
            # table[num] = table.get(num, 0) + 1 # 如果num在table中已经存在，则将其值加1，否则将其值设为1
            
        result = set() # use a set to store the intersection elements
        for num in nums2:
            if num in table:
                result.add(num)
        return list(result) # convert the set to a list and return it
    
    
class Solution:
    def intersection(self, nums1: list[int], nums2: list[int]) -> list[int]:
        return list(set(nums1) & set(nums2))
        
        
```

```python
# happy number
class Solution:
    def isHappy(self, n: int) -> bool:
        seen = set() # 检查是否重复的
        while n not in seen: # 如果没有重复的加入， 有的话直接返回错误
            seen.add(n)
            
            new_num = 0 
            n_str = str(n)  # 拆出来 乘方
            for i in n_str:
                new_num = new_num + int(i) ** 2
            if new_num == 1:
                return True
            else:
                n = new_num
        return False
```

```python
# 两数之和

class Soution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        records = {}
        for index, value in enumerate(nums):
            if target - value in records:
                return [records[target-value], index]
            records[value] = index
        return [] # dont find anything that satisfy the condition, return empty list 
    
    
```

```python
# 四数字相加 Ⅱ
class Solution:
    def fourSumCount(self, nums1: list[int], nums2: list[int], nums3: list[int], nums4: list[int]) -> int:
        # use 字典存储 nums1 和 nums2的元素 及其和
        hashmap = {}

        for n1 in nums1:
            for n2 in nums2:
                if n1 + n2 in hashmap:
                    hashmap[n1 + n2] += 1
                else:
                    hashmap[n1 + n2] = 1
        
        # 如果 - (n1 + n2) 存在于num3 和 nums4中， 存入结果
        count = 0
        for n3 in nums3:
            for n4 in nums4:
                key = - (n3 + n4)
                if key in hashmap:
                    count += hashmap[key]
        return count
    
    

# hash map 存进去

# 如果 有 +1 

# 如果没有 = 1

# 计数

# 第二个数组找， 如果在hashmap里面， cout = count + hashmap[key]

# return coount
```

```python
# 赎金信

from collections import Counter

class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        counts = {}
        for char in magazine:
            if char in counts:
                counts[char] += 1
            else:
                counts[char] = 1
        
        for char in ransomNote:
            if char not in counts or counts[char] == 0:
                return False
            counts[char] -= 1
        return True


class Solution:
    from collections import Counter
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        return not Counter(ransomNote) - Counter(magazine) 
    # 做减法的时候 只有数量大于0的才会留下来 所以如果结果为空
```

```python
# 三数之和     
# 一个迭代    左右指针找


class Solution: # 双指针
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        if len(nums) < 3:
            return []
        
        result = []
        nums.sort() # 排序
        for i in range(len(nums)):
            # if the first number is bigger than 0, return
            if nums[i] > 0:
                return result
        
            # skip the same number to avoid duplicates
            if i > 0 and nums[i] == nums[i - 1]:
                continue  # 对i 去重
        
            # 双指针
            left = i + 1
            right = len(nums) - 1
            
            while left < right:
                sum_ = nums[i] + nums[left] + nums[right]
                
                
                if sum_ < 0:
                    left += 1
                elif sum_ > 0:
                    right -= 1
                else:
                    result.append([nums[i], nums[left], nums[right]])
                    
                    # check if there has the same element, 
                    # if so skip it to avoid duplicates
                    while right > left and nums[right] == nums[right - 1]: # right 去重
                        right -= 1 
                    while right > left and nums[left] == nums[left + 1]: # left 去重
                        left += 1
                        
                    right -= 1
                    left += 1
                    
        return result                   
```
